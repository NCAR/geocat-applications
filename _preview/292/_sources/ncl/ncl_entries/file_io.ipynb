## Functions

### addfile
NCL's [`addfile`](https://www.ncl.ucar.edu/Document/Functions/Built-in/addfile.shtml) opens a data file from supported file formats

#### Grab and Go

- netCDF, HDF5, and HDF-EOS5

````python
import geocat.datafiles as gcd
import xarray as xr

data_filepath = gcd.get("netcdf_files/1994_256_FSD.nc")
netCDF_datafile = xr.open_dataset(data_filepath)
netCDF_datafile
````

#### Grab and Go

- HDF and HDF-EOS

By default, `xarray` supports HDF5 files, but can be updated to a specified engine to work with additional files formats

````python
import geocat.datafiles as gcd
import xarray as xr

data_filepath = gcd.get("hdf_files/avhrr.hdf")
hdf_datafile = xr.open_dataset(data_filepath, engine="netcdf4")
hdf_datafile
````

#### Additional file formats:

```{admonition} Important Note
GRIB files need the additional `cfgrib` engine, which can be installed alongside Xarray:
```
```
conda install -c conda-forge cfgrib eccodes
```
```
import geocat.datafiles as gcd
import xarray as xr

grib_datafile = xr.open_dataset(
    gcd.get("grib_files/ST4.2002030112.06h.grb"), engine="cfgrib"
)
grib_datafile
```

```{admonition} Important Note
Shapefiles can be read via the `geopandas` `read_file()` function. When using Shapefiles, if the index file (`.shx`) is missing, it can be repaired with `gdal` by setting `SHAPE_RESTORE_SHX` to `YES` before reading the file
```
```
conda install -c conda-forge gdal geopandas
```
```
import geocat.datafiles as gcd
import geopandas as gpd

from osgeo import gdal

gdal.SetConfigOption('SHAPE_RESTORE_SHX', 'YES')

shp_datafile = gpd.read_file(gcd.get("shape_files/states.shp"))
```
